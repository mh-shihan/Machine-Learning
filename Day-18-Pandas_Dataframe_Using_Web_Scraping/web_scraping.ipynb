```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import numpy as np
```

```python
webpage=requests.get('https://books.toscrape.com/catalogue/page-1.html').text
```

```python
soup=BeautifulSoup(webpage,'lxml')
```

```python
products = soup.find_all('li', class_='col-xs-6')
```

```python
data = []
```

```python
for product in products:
  name = product.find('h3').text
  price = product.find('p', class_='price_color').text
  rating = product.find('p', class_='star-rating')['class'][1]
  availability = product.find('p', class_='instock availability').text.strip()

  data.append({
    'Name': name,
    'Price': price,
    'Rating': rating,
    'Availability': availability
})
```

```python

base_url = "https://books.toscrape.com/catalogue/page-{}.html"

data = []

for page in range(1, 51):   # 1 to 50 pages
    url = base_url.format(page)
    webpage = requests.get(url).text
    soup = BeautifulSoup(webpage, 'lxml')

    # print(soup.prettify())
    products = soup.find_all('li', class_='col-xs-6')
    for product in products:
      name = product.find('h3').text
      price = product.find('p', class_='price_color').text
      rating = product.find('p', class_='star-rating')['class'][1]
      availability = product.find('p', class_='instock availability').text.strip()

      data.append({
          'Name': name,
          'Price': price,
          'Rating': rating,
          'Availability': availability
      })
```

```python

# Convert to DataFrame
df = pd.DataFrame(data)
```

```python
# Save to CSV
df.to_csv('/content/drive/MyDrive/Colab/Datasets/books_data.csv', index=False)

print("✅ Data saved to books_data.csv")
```
